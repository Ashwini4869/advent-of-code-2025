```python
# Function to read file
def read_file(filename:str):
    data = []
    with open(filename, "r") as file:
        for line in file:
            data.append(line.strip("\n"))
    return data
```

```python
# Replace the required indexes with "|" symbol
def replace_str_index(row, index):
    replacing_char = "|"
    for idx in index:
        row = row[:idx] + replacing_char + row[idx+1:]
    return row
```

```python
# Count split function for part-one
data = read_file("input.txt")

def count_split(data):
    rows = len(data)
    cols = len(data[0])
    split_count = 0
    for i in range(rows - 1):
        if i == 0:
            src_idx = (data[0]).find("S")
            data[1] = replace_str_index(row=data[1], index=[src_idx])
        elif i % 2 != 0:
            idxs_to_replace = []
            for j in range(cols):
                if data[i][j] == "|" and data[i+1][j] == "^":
                    idx1 = j - 1
                    idxs_to_replace.append(idx1)
                    idx2 = j + 1
                    idxs_to_replace.append(idx2)
                    split_count += 1
                elif data[i][j] == "|" and data[i+1][j] == ".":
                    idxs_to_replace.append(j) 
            data[i+1] = replace_str_index(row=data[i+1], index=idxs_to_replace)
        elif i % 2 == 0:
            idxs_to_replace = []
            for j in range(cols):
                if data[i][j] == "|":
                    idxs_to_replace.append(j)
            data[i+1] = replace_str_index(row=data[i+1], index=idxs_to_replace)
    # print(data)
    return split_count

count_split(data)
```
